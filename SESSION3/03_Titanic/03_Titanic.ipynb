# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
df.describe("")
```

```python
# Median cho Age và Fare
median_age = df['age'].median()
median_fare = df['fare'].median()
print("Median Age:", median_age)
print("Median Fare:", median_fare)
```

```python
# Mode cho Age và Fare
mode_age = df['age'].mode()[0]
mode_fare = df['fare'].mode()[0]
print("Mode Age:", mode_age)
print("Mode Fare:", mode_fare)
```

```python
# TODO
# Mean cho Age và Fare
mean_age = df['age'].mean()
mean_fare = df['fare'].mean()
print("Mean Age:", mean_age)
print("Mean Fare:", mean_fare)
```

## Group 2 — Dispersion

```python
# TODO
var_age = df['age'].var()
std_age = df['age'].std()
range_age = df['age'].max() - df['age'].min()

print("Variance (Age):", var_age)
print("Standard Deviation (Age):", std_age)
print("Range (Age):", range_age)

var_fare = df['fare'].var()
std_fare = df['fare'].std()
range_fare = df['fare'].max() - df['fare'].min()

print("Variance (Fare):", var_fare)
print("Standard Deviation (Fare):", std_fare)
print("Range (Fare):", range_fare)
```

## Group 3 — Location and Shape

```python
# TODO
quantiles_age = df['age'].quantile([0.25, 0.5, 0.75])
skewness_age = df['age'].skew()
kurtosis_age = df['age'].kurtosis()

print("Quantiles (Age):\n", quantiles_age)
print("Skewness (Age):", skewness_age)
print("Kurtosis (Age):", kurtosis_age)
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python

# TODO
# Tỷ lệ sống sót theo hạng vé
survival_by_class = df.groupby('class')['survived'].mean().sort_values(ascending=False)
print("Tỷ lệ sống sót theo hạng vé:\n", survival_by_class)

# Chênh lệch giữa cao nhất và thấp nhất
diff_class = survival_by_class.max() - survival_by_class.min()
print("Chênh lệch giữa hạng cao nhất và thấp nhất:", diff_class)
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
# Tỷ lệ sống sót theo giới tính
survival_by_sex = df.groupby('sex')['survived'].mean()
print("Tỷ lệ sống sót theo giới tính:\n", survival_by_sex)

# TODO
# So sánh độ chênh lệch
diff_sex = survival_by_sex.max() - survival_by_sex.min()
print("Chênh lệch giữa giới tính:", diff_sex)

```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
# Tính tương quan giữa fare và survived
corr_fare_survived = df[['fare','survived']].corr().iloc[0,1]
print("Tương quan giữa Fare và Survived:", corr_fare_survived)

```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
# Tỷ lệ sống sót theo family_size
survival_by_family = df.groupby('family_size')['survived'].mean()
print("Tỷ lệ sống sót theo family_size:\n", survival_by_family)

```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO
# Tỷ lệ sống sót theo embark_town
survival_by_embark = df.groupby('embark_town')['survived'].mean().sort_values(ascending=False)
print("Tỷ lệ sống sót theo cảng lên tàu:\n", survival_by_embark)

```
